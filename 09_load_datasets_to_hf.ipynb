```python
from datasets import (
    load_dataset,
    DatasetDict,
)
```

```python
# ==================================================
# Configuration
# ==================================================

DATASET_NAME = "stanfordnlp/imdb"
HF_DATASET = "YOUR_USERNAME/imdb-prepared"
LOCAL_PATH = "./imdb-prepared"
TRAIN_SIZE = 1000
TEST_SIZE = 200
```

```python

# ==================================================
# 1. Load original dataset
# ==================================================

print("Loading dataset...")

dataset = load_dataset(
    DATASET_NAME
)

print(dataset)
```

```python
# ==================================================
# 2. Clean dataset
# ==================================================

def clean_example(example):

    example["text"] = example["text"].strip()

    return example


dataset = dataset.map(clean_example)
```

```python
# ==================================================
# 3. Create smaller dataset
# ==================================================

train_dataset = dataset["train"].select(
    range(TRAIN_SIZE)
)

test_dataset = dataset["test"].select(
    range(TEST_SIZE)
)


prepared_dataset = DatasetDict({
    "train": train_dataset,
    "test": test_dataset,
})


print("\nPrepared dataset:")

print(prepared_dataset)
```

```python
# ==================================================
# 4. Save locally
# ==================================================

print("\nSaving locally...")

prepared_dataset.save_to_disk(
    LOCAL_PATH
)

print(
    f"Saved to {LOCAL_PATH}"
)
```

```python

# ==================================================
# 5. Upload to Hugging Face
# ==================================================

print("\nUploading to Hugging Face...")

prepared_dataset.push_to_hub(
    HF_DATASET
)

print(
    f"Uploaded to: {HF_DATASET}"
)
```
