```python
import random
from datasets import Dataset
```

```python
# ============================================================
# Configuration
# ============================================================

NUM_EXAMPLES = 1000
OUTPUT_FILE = "synthetic_data.jsonl"
HF_DATASET = "worldboss/python-instructions"
```

```python
# ============================================================
# Synthetic data generators
# ============================================================

def generate_python_example():

    examples = [

        {
            "instruction": "Explain what a Python list is.",
            "input": "",
            "output": (
                "A Python list is an ordered, mutable collection "
                "that can contain multiple values."
            ),
        },

        {
            "instruction": "Explain what a Python dictionary is.",
            "input": "",
            "output": (
                "A Python dictionary stores data as key-value pairs. "
                "It is mutable and keys must be hashable."
            ),
        },

        {
            "instruction": "Explain a Python function.",
            "input": "",
            "output": (
                "A Python function is a reusable block of code "
                "defined with the def keyword."
            ),
        },

        {
            "instruction": "Explain a Python class.",
            "input": "",
            "output": (
                "A Python class is a blueprint for creating objects "
                "that contain attributes and methods."
            ),
        },

        {
            "instruction": "What does self mean in Python?",
            "input": "",
            "output": (
                "self refers to the current instance of a class. "
                "It allows instance methods to access instance attributes."
            ),
        },

        {
            "instruction": "Explain inheritance in Python.",
            "input": "",
            "output": (
                "Inheritance allows a class to reuse attributes and "
                "methods from another class."
            ),
        },

        {
            "instruction": "Explain polymorphism in Python.",
            "input": "",
            "output": (
                "Polymorphism allows objects of different classes "
                "to respond to the same method interface."
            ),
        },

    ]

    return random.choice(examples)
```

```python
# ============================================================
# Mathematical examples
# ============================================================

def generate_math_example():

    a = random.randint(1, 100)
    b = random.randint(1, 100)

    operation = random.choice(
        ["add", "subtract", "multiply"]
    )

    if operation == "add":

        answer = a + b

        instruction = "Calculate the sum of two numbers."

        input_text = f"{a} + {b}"

        output = f"The answer is {answer}."

    elif operation == "subtract":

        answer = a - b

        instruction = "Calculate the difference between two numbers."

        input_text = f"{a} - {b}"

        output = f"The answer is {answer}."

    else:

        answer = a * b

        instruction = "Calculate the product of two numbers."

        input_text = f"{a} * {b}"

        output = f"The answer is {answer}."

    return {
        "instruction": instruction,
        "input": input_text,
        "output": output,
    }
```

```python
# ============================================================
# String examples
# ============================================================

def generate_string_example():

    words = [
        "python",
        "machine learning",
        "artificial intelligence",
        "data science",
        "hugging face",
        "deep learning",
    ]

    word = random.choice(words)

    operation = random.choice(
        ["uppercase", "lowercase", "length"]
    )

    if operation == "uppercase":

        answer = word.upper()

        instruction = (
            "Convert the following text to uppercase."
        )

        output = answer

    elif operation == "lowercase":

        answer = word.lower()

        instruction = (
            "Convert the following text to lowercase."
        )

        output = answer

    else:

        answer = len(word)

        instruction = (
            "Calculate the number of characters in the text."
        )

        output = f"The text contains {answer} characters."

    return {
        "instruction": instruction,
        "input": word,
        "output": output,
    }
```

```python
# ============================================================
# Generate dataset
# ============================================================

data = []


for _ in range(NUM_EXAMPLES):

    category = random.choice(
        [
            "python",
            "math",
            "string",
        ]
    )

    if category == "python":

        example = generate_python_example()

    elif category == "math":

        example = generate_math_example()

    else:

        example = generate_string_example()

    example["category"] = category

    data.append(example)
```

```python
data
```

```python
# ============================================================
# Convert to Hugging Face Dataset
# ============================================================

dataset = Dataset.from_list(data)
```

```python
# ============================================================
# Inspect dataset
# ============================================================

print(dataset)
print("\nFirst example:")
print(dataset[0])


print("\nFirst 5 examples:")

for example in dataset.select(range(5)):
    print(example)
```

```python
# ============================================================
# Save locally
# ============================================================

dataset.save_to_disk("./synthetic-python-dataset")


print("\nDataset saved locally.")
```

### Create a Qwen3 training format

### Instruction:
Explain what a Python class is.

### Input:


### Response:
A Python class is a blueprint for creating objects...

```python
def format_example(example):

    if example["input"]:

        prompt = f"""### Instruction:
{example["instruction"]}

### Input:
{example["input"]}

### Response:
{example["output"]}"""

    else:
        prompt = f"""### Instruction:
{example["instruction"]}

### Response:
{example["output"]}"""

    return {
        "text": prompt
    }
```

```python
formatted_dataset = dataset.map(format_example)

print(formatted_dataset["train"][2]["text"])
```
